 # Task 1:

```python
side1 = float(input("Enter side 1: "))
side2 = float(input("Enter side 2: "))
side3 = float(input("Enter side 3: "))

if side1 == side2 and side2 == side3:
    print("This is an equilateral triangle")
elif side1 == side2 or side1 == side3 or side2 == side3:
    print("This is an isosceles triangle")
else:
    print("This is a scalene triangle")
```

# Task 2:

```python
student_grades = []

for i in range(5):
    grade = float(input("Enter grade for student " + str(i+1) + ": "))
    student_grades.append(grade)

print(student_grades)

total = 0
for g in student_grades:
    total = total + g
average = total / len(student_grades)

print("Class average:", average)
print("Minimum grade:", min(student_grades))
print("Maximum grade:", max(student_grades))

at_risk = 0
for g in student_grades:
    if g < 65:
        at_risk += 1

print(at_risk, "students are at risk in the course.")
```

# Task 3:

```python
# A
mult_of_2 = list(range(2, 101, 2))
print("Multiples of 2:", mult_of_2)

# B
mult_of_3 = list(range(3, 101, 3))
print("Multiples of 3:", mult_of_3)
```

```python
# C
mult_of_both = [x for x in mult_of_2 if x in mult_of_3]
print("Multiples of both:", mult_of_both)
```

```python
# D
mult_of_either = []
for num in range(1, 101):
    if num in mult_of_2 or num in mult_of_3:
        if num not in mult_of_either:
            mult_of_either.append(num)

print("Multiples of either:", mult_of_either)
print(len(mult_of_either))
```

# Task 4: 

```python
shopping_list = []

count = 0
while count < 7:
    item = input("Enter an item (or type done to stop): ")
    if item.lower() == "done":
        break
    shopping_list.append(item)
    count = count + 1

# B
print("Total number of items:", len(shopping_list))

# C
print("Reverse order:", shopping_list[::-1])

# D
alpha_list = sorted(shopping_list)
print("Alphabetical order:", alpha_list)

# E 
already_said = []
for i in range(len(shopping_list)):
    for j in range(i+1, len(shopping_list)):
        if shopping_list[i] == shopping_list[j] and shopping_list[i] not in already_said:
            print(shopping_list[i], "is on your list more than once!")
            already_said.append(shopping_list[i])

if len(already_said) == 0:
    print("No items show up more than once")
```
