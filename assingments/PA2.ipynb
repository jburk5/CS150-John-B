## Task 2: Distance Conversion
Put a distance in miles in `miles_distance`, convert it to kilometers with kilometers = miles * 1.60934, and print both.

```python
miles_distance = 26.2  # marathon
kilometers = miles_distance * 1.60934

print("The distance is", miles_distance, "miles, which is the equivalent of", round(kilometers, 2), "kilometers.")
```

## Task 3: Working with Running Times
Make two variables `run1` and `run2` with run times in minutes, then print:

- B1. the total time (add them)
- B2. the improvement (run1 - run2)
- B3. run2 times 2, like doing it twice
- B4. the average of both runs

```python
run1 = 24
run2 = 21.5

# B1
total = run1 + run2
print("Total time running:", total, "minutes")

# B2
improvement = run1 - run2
print("Improvement:", improvement, "minutes")

# B3
twice = run2 * 2
print("Run 2 done twice:", twice, "minutes")

# B4
average = (run1 + run2) / 2
print("Average run:", average, "minutes")
```

## Task 4: String Manipulation
Store a famous quote in `quote`. Print its length, print it in uppercase and lowercase, print it with the leading/trailing whitespace removed, and print it split into a list of words.

```python
# Wayne Gretzky quote, added spaces on the ends so strip() has something to remove
quote = "  You miss 100% of the shots you don't take.  "

print(len(quote))
print(quote.upper())
print(quote.lower())
print(quote.strip())
print(quote.split())
```

## Task 5: User Input and Operations
Ask the user for the lengths of the two legs of a right triangle, find the hypotenuse with the Pythagorean theorem c = sqrt(a^2 + b^2), and print it rounded to the nearest hundredth.

```python
import math

a = float(input("Enter the length of the first leg: "))
b = float(input("Enter the length of the second leg: "))

c = math.sqrt(a**2 + b**2)

print("The hypotenuse of your triangle is", round(c, 2), "units.")
```
